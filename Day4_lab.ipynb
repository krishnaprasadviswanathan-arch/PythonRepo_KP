```python
marks=[78, 85, 62, 90, 55]
print(marks)
print(len(marks))
print(marks[0])
print(marks[-1])
print(marks[1:3])
marks[2] =70  #changign the mark
print(marks)
marks[1:3] =[88,92]  #replacing the mark at speified index
print(marks)
#adding mark using append
marks.append(76)
print(marks)
#insert 60 at index1
marks.insert(1,60)
print(marks)
#extend the list
marks.extend([81,84])
print(marks)
#Remove the mark 55
marks.remove(55)
print(marks)
# Remove the last element using pop()
marks.pop(-1)
print(marks)
```

```python
#Loop through the list and print each mark:
for i in marks:
    print(i)
 
```

```python
#Sort passed_marks in ascending order
marks.sort()
print(marks)
# sort it in descending order since the above mark is sorted we can user just reverse as it will print in reverse which is descending
marks.sort(reverse=True)
print(marks) 
```
