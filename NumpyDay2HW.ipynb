```python
# Generate 5 random values between 0 and 1 representing normalized scores using
import numpy as np
np1 = np.random.rand(5)
print(np1)
# Generate 5 random values from a normal distribution using 
np2 = np.random.randn(5)
print(np2)
```

```python
# Create a 3×4 array of random integers between 35 and 100 using
import numpy as np
np1 = np.random.randint(35,100,(3,4))
print(np1)
```

```python
# Given a list of student IDs Randomly select 2 students using
import numpy as np
np1=np.array([101, 102, 103, 104, 105])
print(np.random.choice(np1,2))          # (np.random.choice(arr, size=2))
```

```python
# Fix the Random Output
# Set a random seed value 10
import numpy as np
np.random.seed(10)
# Generate 3 random values using
# Observe that output remains the same every time
np1 = np.random.rand(3)
print(np1)
```

```python
# Basic Indexing & Slicing
# Using the array [10, 20, 30, 40, 50, 60]
import numpy as np
np1 = np.array([10, 20, 30, 40, 50, 60])
# Print the 3rd element
print(np1[2])
# Print elements from index 1 to 4
print(np1[1:5])
# Print the last element
print(np1[-1])
# Print every second element
print(np1[::2])
# Reverse the array
print(np1[::-1])
```
