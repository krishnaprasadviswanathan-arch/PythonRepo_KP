```python
#Perform: Total marks (addition), Difference (term2 − term1),Increase percentage using multiplication by 1.05
import numpy as np
marks_term1 = np.array([60, 70, 80, 90])
marks_term2 = np.array([65, 75, 85, 95])
Total = marks_term1 + marks_term2
print(Total)
diff = marks_term2 - marks_term1
print(diff)
mark_increase_term1 = marks_term1 * 1.05
mark_increase_term2 = marks_term2 * 1.05
print(mark_increase_term1)
print(mark_increase_term2)
#From the total marks array: Find marks greater than 150 & Find marks between 140 and 170
print(Total[Total>150])
print(Total[(Total>=140) & (Total<=170)])
#From the total marks array: Find Sum,Mean,Maximum,Minimum,Standard deviation
print(Total.sum())
print(Total.mean())
print(Total.max())
print(Total.min())
print(Total.std())

```

```python
#Create a 2×3 array of marks: Column-wise sum and Row-wise mean
import numpy as np
arr = np.array([[70, 80, 90],
                [60, 75, 85]])
print(arr)
print(arr.sum(axis=0))
print(arr.sum(axis=1))
```
