```python
# Create a 3×4 array of zeros
import numpy as np
npzero = np.zeros((3,4))
print(npzero)
#Create a 2×3 array of ones
npones=np.ones((2,3))
print(npones)
#Create a 3×3 array filled with value 5
npfull=np.full((3,3),5)
print(npfull)
# Create a 4×4 identity matrix
npiden = np.eye((4))
print(npiden)
# Create a diagonal matrix with values
npdia = np.diag([10, 20, 30, 40])
print(npdia)
```
