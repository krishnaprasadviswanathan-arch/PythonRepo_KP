```python
import matplotlib.pyplot as plt
plt.figure(figsize=(12,4))
fig,ax=plt.subplots(2,1,sharex=True)
plt.xlabel("Month")
plt.ylabel("Sales")
x=[1, 2, 3, 4]
y1=[200, 250, 300, 350]
y2=[180, 220, 260, 310]
ax[0].plot(x,y1,linestyle="-",color="green",marker="o")
ax[1].plot(x,y2,linestyle="--",color="red",marker="s")
plt.show()
```
