```python
# Student Details using Dictionary
Stud_Dict = [{"name":"Krishna","age":19,"course":"Science"},
             {"name":"Ramesh","age":25,"course":"Humanities"},
             {"name":"Rakesh","age":22,"course":"Medicine"}]
print(type(Stud_Dict))
# Print all student names
for student in Stud_Dict :
    print (student["name"])
# Print each student’s name with age and course
for student in Stud_Dict :
    print("student details = ",student.values())
# Update one student’s course
for student in Stud_Dict :
    if student["name"] == "Krishna":
        student["age"] = 23
print(Stud_Dict)
# Remove one student record
for student in Stud_Dict :
    if student["name"] == "Ramesh":
        Stud_Dict.remove(student)

print(Stud_Dict)        
```
