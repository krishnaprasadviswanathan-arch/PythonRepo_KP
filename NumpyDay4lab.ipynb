```python
#Create a NumPy array of student marks - Pass" if marks ≥ 40 , "Fail" otherwise , Print the result
import numpy as np
np1 = np.array([35, 55, 75, 40, 90])
print(np.where(np1>=40,"Pass","Fail"))
#Assign grades: Marks < 40 → "Fail",Marks < 60 → "C",Marks < 80 → "B",Else → "A"
conditions=[np1 < 40,np1 < 60,np1 < 80]
choices=["Fail","C","B"]
print(np.select(conditions,choices,default="A"))





```

```python
#Create an array with duplicate marks: [50, 60, 50, 70, 60, 80] and remove duplicate
import numpy as np
marks = np.array([50, 60, 50, 70, 60, 80])
print(np.unique(marks))
#Check if any student has failed (marks < 40) using any
print(np.any(marks<40))
#Check if all students passed (marks ≥ 40)
print(np.all(marks>40))
```
