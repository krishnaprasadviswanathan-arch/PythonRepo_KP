```python
# Create a NumPy array of sensor readings: 
import numpy as np
sensor_read = np.array([25.5, 26.0, np.nan, 27.8, np.inf, 26.4])
print(np.isnan(sensor_read))
print(np.isinf(sensor_read))
sensor_read[np.isnan(sensor_read)] = 0
print(sensor_read)
sensor_read[np.isinf(sensor_read)] = 0
print(sensor_read)
```

```python
#to remove all the unwanted values
import numpy as np
sensor_read = np.array([25.5, 26.0, np.nan, 27.8, np.inf, 26.4])
clean_arr = sensor_read[~np.isnan(sensor_read)]
clean_arr1 = clean_arr[~np.isinf(clean_arr)]
print(clean_arr1)
```
