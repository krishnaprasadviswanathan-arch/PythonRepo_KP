```python
# Weekly Website Visitors Trend
import matplotlib.pyplot as plt
plt.figure(figsize=(8,4))
plt.style.use('seaborn-bright')
x=[1, 2, 3, 4]
y=[120, 150, 130, 180]
plt.plot(x,y,linestyle="-",marker="o",color="blue")
plt.xlabel("Day")
plt.ylabel("No:of Visitors")
#plt.savefig("Websitevisi.png")
plt.show()
```
